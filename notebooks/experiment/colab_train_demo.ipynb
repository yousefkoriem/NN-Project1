```python
(x_train,y_train) = pickle.load(open('data/processed/imdb_train.pkl', 'rb'))
(x_test,y_test) = pickle.load(open('data/processed/imdb_test.pkl', 'rb'))
```

```python
x_train.shape, y_train.shape, x_test.shape, y_test.shape
```

```python
model = models.load_model('models/architecture/untrained_demo_model.keras')
```

```python
erlystp = keras.callbacks.EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
```

```python
history = model.fit(x_train, y_train, epochs=64, batch_size=64, validation_split=0.2, callbacks=[erlystp])
```

```python
score = model.evaluate(x_test, y_test)
```

```python
history_df = pd.DataFrame(history.history)
score_df = pd.DataFrame(score, index=['loss', 'accuracy','f1_score', 'precision', 'recall']).T
```

```python
model.save('trained_demo_model.keras')
```

```python
joblib.dump(score_df, 'score.h5')
joblib.dump(history_df, 'history.h5')
```
